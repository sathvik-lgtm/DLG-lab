```python
import numpy as np
from sklearn.datasets import make_moons
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input
from tensorflow.keras.optimizers import SGD

X, y = make_moons(n_samples=400, noise=0.20, random_state=1)
y = y.reshape(-1, 1).astype(float)
X = (X - X.mean(0)) / X.std(0)

def create_model():
    model = Sequential([
        Input(shape=(2,)),
        Dense(16, activation='relu'),
        Dense(16, activation='relu'),
        Dense(1, activation='sigmoid')
    ])

    model.compile(
        loss='binary_crossentropy',
        optimizer=SGD(learning_rate=0.5),
        metrics=['accuracy']
    )

    return model

batch_model = create_model()

batch_model.fit(
    X,
    y,
    epochs=200,
    batch_size=len(X),
    verbose=0
)

batch_loss, batch_acc = batch_model.evaluate(X, y, verbose=0)

sgd_model = create_model()

sgd_model.fit(
    X,
    y,
    epochs=200,
    batch_size=16,
    verbose=0
)

sgd_loss, sgd_acc = sgd_model.evaluate(X, y, verbose=0)
print(f"Batch GD Accuracy: {batch_acc:.4f}")
print(f"Mini-Batch SGD Accuracy: {sgd_acc:.4f}")
```
